## Building differentiable pipelines

Calling `.apply()` as below runs a Tesseract as a plain function. To compose
Tesseracts into a *differentiable* pipeline—so you can take gradients through
the whole chain—use the framework bindings instead of calling `.apply()`
directly:

- **[Tesseract-JAX](https://github.com/pasteurlabs/tesseract-jax)** runs
  Tesseracts inside JAX programs, with full support for `jit`, `grad`, and
  other transformations.
- **[Tesseract-Torch](https://github.com/pasteurlabs/tesseract-torch)** wraps
  Tesseracts as differentiable PyTorch operations, with forward- and
  reverse-mode automatic differentiation.

These pair naturally with the `jax` and `pytorch` recipes
(`make new <mytess> -- --recipe jax`), which scaffold differentiable
endpoints for your component.
